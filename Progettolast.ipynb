**WINE DATASET**


```python
import numpy as np
np.random.seed(77)

train_indices = []
val_indices = []
test_indices = []

```

```python
for class_value in [0, 1]:
    class_indices = y_wine[y_wine == class_value].index.to_numpy()
    np.random.shuffle(class_indices)

    n_samples = len(class_indices)
    train_max = int(n_samples * 0.60)
    val_max = int(n_samples * 0.80)

    train_indices.extend(class_indices[:train_max])
    val_indices.extend(class_indices[train_max:val_max])
    test_indices.extend(class_indices[val_max:])
```

```python
np.random.shuffle(train_indices)
np.random.shuffle(val_indices)
np.random.shuffle(test_indices)
```

```python
X_train = X_wine.loc[train_indices].reset_index(drop=True)
y_train = y_wine.loc[train_indices].reset_index(drop=True)
X_val = X_wine.loc[val_indices].reset_index(drop=True)
y_val = y_wine.loc[val_indices].reset_index(drop=True)
X_test = X_wine.loc[test_indices].reset_index(drop=True)
y_test = y_wine.loc[test_indices].reset_index(drop=True)
```

```python
print(f"Train set: {len(X_train)} rows")
print(f"Val set:   {len(X_val)} rows")
print(f"Test set:  {len(X_test)} rows")
```

```python
# DataFrame in array NumPy
X_train2 = X_train.to_numpy()
X_val2 = X_val.to_numpy()
X_test2 = X_test.to_numpy()
```

```python
mean_train = X_train2.mean(axis=0)
std_train = X_train2.std(axis=0)
std_train[std_train == 0] = 1e-11

X_train_scaled = (X_train2 - mean_train) / std_train
X_val_scaled = (X_val2 - mean_train) / std_train
X_test_scaled = (X_test2 - mean_train) / std_train
```

```python
from sklearn.linear_model import LogisticRegression

# Logistic
lr_model = LogisticRegression(max_iter=5000)
lr_model.fit(X_train_scaled, y_train)
y_val_probs_lr = lr_model.predict_proba(X_val_scaled)[:, 1]
y_val_pred_lr = (y_val_probs_lr >= 0.5).astype(int)
```

```python
from sklearn.ensemble import RandomForestClassifier

# Random Forest
rf_model = RandomForestClassifier(n_estimators=200, random_state=15, n_jobs=-1)
rf_model.fit(X_train_scaled, y_train)
y_val_probs_rf = rf_model.predict_proba(X_val_scaled)[:, 1]
y_val_pred_rf = (y_val_probs_rf >= 0.5).astype(int)
```

```python
def accuracy(y_true, y_pred):
    return np.mean(y_true == y_pred)

def brier_score(y_true, y_probs):
    return np.mean((y_probs - y_true) ** 2)

def log_loss(y_true, y_probs):
    eps = np.finfo(float).eps
    y_probs = np.clip(y_probs, eps, 1 - eps)
    return -np.mean(y_true * np.log(y_probs) + (1 - y_true) * np.log(1 - y_probs))
```

```python

print("--- Random Forest (Validation) ---")
print(f"Accuracy:    {accuracy(y_val, y_val_pred_rf):.4f}")
print(f"Brier Score: {brier_score(y_val, y_val_probs_rf):.4f}")
print(f"Log-Loss:    {log_loss(y_val, y_val_probs_rf):.4f}")
```

```python
import matplotlib.pyplot as plt

def plot_reliability_diagram(y_true, y_probs, title="", n_bins=10):
    y_true = np.asarray(y_true)
    y_probs = np.asarray(y_probs)
    bin_edges = np.linspace(0, 1, n_bins + 1)

    frac_pos, mean_pred = [], []
    for i in range(n_bins):
        if i == n_bins - 1:
            in_bin = (y_probs >= bin_edges[i]) & (y_probs <= bin_edges[i + 1])
        else:
            in_bin = (y_probs >= bin_edges[i]) & (y_probs < bin_edges[i + 1])
        if in_bin.sum() > 0:
            frac_pos.append(y_true[in_bin].mean())
            mean_pred.append(y_probs[in_bin].mean())

    plt.plot([0, 1], [0, 1], "--")
    plt.plot(mean_pred, frac_pos, marker="o")
    plt.xlabel("Predicted average probability")
    plt.ylabel("Actual frequency")
    plt.title(title)
    plt.show()
```

```python
plot_reliability_diagram(y_val, y_val_probs_lr, title="Logistic Regression")
```

```python
plot_reliability_diagram(y_val, y_val_probs_rf, title="Random Forest")
```

```python
def logit(p):
    p = np.clip(p, 1e-3, 1 - 1e-3)
    return np.log(p / (1 - p))

def platt_fit(f, y, lr=0.1, iterations=5000):
    y = np.asarray(y)
    n_pos = y.sum()
    n_neg = len(y) - n_pos
    t = np.where(y == 1, (n_pos + 1) / (n_pos + 2), 1 / (n_neg + 2))

    A, B = -1.0, 0.0
    for _ in range(iterations):
        p = 1 / (1 + np.exp(A * f + B))
        A -= lr * np.mean((t - p) * f)
        B -= lr * np.mean(t - p)
    return A, B

def platt_predict(A, B, f):
    return 1 / (1 + np.exp(A * f + B))
```

```python
y_test_probs_lr = lr_model.predict_proba(X_test_scaled)[:, 1]
y_test_probs_rf = rf_model.predict_proba(X_test_scaled)[:, 1]
```

```python
A_lr, B_lr = platt_fit(logit(y_val_probs_lr), y_val)
A_rf, B_rf = platt_fit(logit(y_val_probs_rf), y_val)
print("A, B (LR):", A_lr, B_lr)
print("A, B (RF):", A_rf, B_rf)
```

```python
y_test_probs_lr_platt = platt_predict(A_lr, B_lr, logit(y_test_probs_lr))
y_test_probs_rf_platt = platt_predict(A_rf, B_rf, logit(y_test_probs_rf))
```

```python
for name, p in [("Basic LR", y_test_probs_lr), ("LR + Platt", y_test_probs_lr_platt), ("Basic RF", y_test_probs_rf), ("RF + Platt", y_test_probs_rf_platt)]: print(f"{name:<12} | Accuracy: {accuracy(y_test, (p >= 0.5).astype(int)):.4f} " f"| Brier: {brier_score(y_test, p):.4f} | Log-loss: {log_loss(y_test, p):.4f}")
```

```python
def isotonic_fit(f, y):
    order = np.argsort(f)
    f = np.asarray(f)[order]
    y = np.asarray(y)[order].astype(float)

    blocks = []
    for fi, yi in zip(f, y):
        blocks.append([yi, 1, fi])
        # se il blocco precedente ha media >= dell'ultimo, li fondiamo
        while len(blocks) > 1 and blocks[-2][0] >= blocks[-1][0]:
            m2, w2, h2 = blocks.pop()
            m1, w1, h1 = blocks.pop()
            blocks.append([(m1 * w1 + m2 * w2) / (w1 + w2), w1 + w2, h2])

    upper = np.array([b[2] for b in blocks])
    means = np.array([b[0] for b in blocks])
    return upper, means

def isotonic_predict(model, f):
    upper, means = model
    idx = np.searchsorted(upper, f)
    return means[np.minimum(idx, len(means) - 1)]
```

```python
iso_lr = isotonic_fit(y_val_probs_lr, y_val)
iso_rf = isotonic_fit(y_val_probs_rf, y_val)
```

```python
y_test_probs_lr_iso = isotonic_predict(iso_lr, y_test_probs_lr)
y_test_probs_rf_iso = isotonic_predict(iso_rf, y_test_probs_rf)
```

```python
for name, p in [("Basic LR", y_test_probs_lr), ("LR + Platt", y_test_probs_lr_platt), ("LR + Isotonic", y_test_probs_lr_iso),
                ("Basic RF", y_test_probs_rf), ("RF + Platt", y_test_probs_rf_platt), ("RF + Isotonic", y_test_probs_rf_iso)]:
    print(f"{name:<14} | Accuracy: {accuracy(y_test, (p >= 0.5).astype(int)):.4f} "
          f"| Brier: {brier_score(y_test, p):.4f} | Log-loss: {log_loss(y_test, p):.4f}")
```

```python
plot_reliability_diagram(y_test, y_test_probs_lr, title="Baisc LR")
plot_reliability_diagram(y_test, y_test_probs_lr_platt, title="LR + Platt")
plot_reliability_diagram(y_test, y_test_probs_lr_iso, title="LR + Isotonic")
```

```python
plot_reliability_diagram(y_test, y_test_probs_rf, title="Basic RF")
plot_reliability_diagram(y_test, y_test_probs_rf_platt, title="RF + Platt")
plot_reliability_diagram(y_test, y_test_probs_rf_iso, title="RF + Isotonic")
```
